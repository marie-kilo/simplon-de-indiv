## 2. Créer et écrire dans un fichier

## 4. Python : modes d'ouvertures d'un fichier (File Modes)

- 'r' : Mode par défaut. Ouvre un fichier pour le lire.
- 'w': Ouvre un fichier pour écrire. Si le fichier n'existe pas, il est créé. (RISQUE d'ECRASÉMENT)
- 'x': Crée un nouveau fichier. Si le fichier existe déjà, l'opération échoue.
- 'a': Ouvre un fichier en mode `append`. S'il n'existe pas, il est créé.
- 't': Mode par défaut, ouvre un fichier `text`
- 'b': Ouvre un fichier de `bytes`
- '+': On indique qu'on veut ouvrir pour lire et écrire

````python
file = open("data/tp_04/file_2.txt", "at+")

for i in range(2):
     file.write(f"Ligne ajoutée ++ line {i}\n")
        
file.close()
````

## 6. Pickle : sauvegarder n'importe quel objet Python

- On peut avec le package `pickle` garder une copie physique de n'importe quel objet Python (peu importe son type)


````python
# Save a dictionary into a pickle file.
import pickle

# On définit un dictionnaire
my_dict = { "Max": 20, "André": 37 }

# On enregistre le dictionnaire au format pickle dans le fichier my_dict.p
# le b dans "wb" indique que l'on écrit des bytes
pickle.dump( my_dict, open("data/tp_04/my_dict.p", "wb" ), protocol=pickle.HIGHEST_PROTOCOL)

# On réouvre ce fichier
# le b dans "rb" indique que l'on lit des bytes
my_dict_opened = pickle.load( open("data/tp_04/my_dict.p", "rb" ) )

assert my_dict == my_dict_opened

# Vérification
print("fichier enregistré", my_dict)
print("fichier ouvert", my_dict_opened)
````

<a id='7'></a>

## 7. La syntaxe `with`

````python
# Syntaxe alternartive

import pickle

my_dict_2 = { "Max": 20, "André": 37 , "Hubert": 19}

# le b dans "wb" indique que l'on écrit des bytes
with open('data/tp_04/my_dict_2.pickle', 'wb') as file:
    pickle.dump(my_dict_2, file, protocol=pickle.HIGHEST_PROTOCOL)

# le b dans "rb" indique que l'on lit des bytes
with open('data/tp_04/my_dict_2.pickle', 'rb') as file:
    my_dict_2_opened = pickle.load(file)

assert my_dict_2 == my_dict_2_opened

# Vérification
print("fichier enregistré", my_dict_2)
print("fichier ouvert", my_dict_2_opened)
````

````python
#2. Ecrire une fonction qui renvoie un dictionnaire avec en clef les extentions présentes et en valeurs le nombre de fichiers avec cette extension, par exemple :
import os

def compter_extensions():
    fichiers = os.listdir()
    dico = {}

    for nom in fichiers:
        # On ignore les dossiers
        if os.path.isfile(nom):
            # On récupère l'extension
            _, ext = os.path.splitext(nom)

            # Si pas d'extension, on peut ignorer ou compter sous "" (à toi de voir)
            if ext != "":
                dico[ext] = dico.get(ext, 0) + 1

    return dico

print(compter_extensions())
        
````

## Exercice 2 - Fichiers et datetimes
- Ecrire une fonction qui pendant 30 secondes crée un fichier toutes les 5 secondes (donc 6 fichiers) : 
    - ce fichier aura comme nom son heure de création
    - Il contiendra le numéro du fichier (1 pour le premier,..., 6 pour le dernier)

````python
# A toi de jouer
import datetime
import time
import os

def creer_fichiers_toutes_les_5_secondes():
    """
    Crée 6 fichiers espacés de 5 secondes pendant 30 secondes.
    Chaque fichier est nommé selon l'heure de création et contient son numéro.
    """
    for i in range(1, 7):  # 6 fichiers
        # Heure actuelle pour le nom du fichier
        maintenant = datetime.datetime.now()
        nom_fichier = maintenant.strftime("%H-%M-%S") + ".txt"

        # Création du fichier
        with open(nom_fichier, "w", encoding="utf-8") as f:
            f.write(f"Fichier numéro {i}")

        print(f"Créé : {nom_fichier}")

        # Attendre 5 secondes avant de créer le suivant
        time.sleep(5)

creer_fichiers_toutes_les_5_secondes()
````

## Exercice 4 - Compter les lignes et les mots d'un fichier

On écrit d'abord un petit texte dans un fichier (donnée fournie), puis tu le relis.

- Écrire une fonction `compter(chemin)` qui ouvre le fichier et renvoie un couple `(nombre_de_lignes, nombre_de_mots)`.
- L'appliquer au fichier `data/tp_04/poeme.txt`.


````python
# Donnée fournie : on crée le fichier à analyser
texte = """Python est un langage clair.
On apprend a lire et a ecrire des fichiers.
Un fichier est une suite de lignes."""
with open("data/tp_04/poeme.txt", "w") as f:
    f.write(texte)

# Votre code ici
def compter(chemin):
    with open(chemin, "r", encoding="utf-8") as f:
        contenu = f.read()

    # Compter les lignes
    lignes = contenu.split("\n")
    nb_lignes = len(lignes)

    # Compter les mots
    mots = contenu.split()
    nb_mots = len(mots)

    return nb_lignes, nb_mots


# Application au fichier demandé
print(compter("data/tp_04/poeme.txt"))
````

## Exercice 5 - Fusionner plusieurs fichiers

On crée d'abord trois fichiers (donnée fournie).

- Écrire une fonction `fusionner()` qui lit tous les fichiers dont le nom commence par `partie_` dans `data/tp_04/`, et écrit leur contenu mis bout à bout dans `data/tp_04/fusion.txt`.
- Indication : `os.listdir("data/tp_04")` donne la liste des noms de fichiers du dossier.


````python
import os

# Donnée fournie : on crée trois fichiers à fusionner
for i in range(1, 4):
    with open(f"data/tp_04/partie_{i}.txt", "w") as f:
        f.write(f"Contenu de la partie {i}\n")

# Votre code ici

import os

def fusionner():
    dossier = "data/tp_04"
    fichiers = os.listdir(dossier)

    # On garde seulement les fichiers qui commencent par "partie_"
    parties = [f for f in fichiers if f.startswith("partie_")]

    # On trie pour avoir partie_1, partie_2, partie_3 dans l'ordre
    parties.sort()

    contenu_total = ""

    # On lit chaque fichier et on ajoute son contenu
    for nom in parties:
        chemin = os.path.join(dossier, nom)
        with open(chemin, "r", encoding="utf-8") as f:
            contenu_total += f.read()

    # On écrit le résultat dans fusion.txt
    with open(os.path.join(dossier, "fusion.txt"), "w", encoding="utf-8") as f:
        f.write(contenu_total)


# On exécute la fonction
fusionner()

#Pour afficher le contenu du fichier fusionné
with open("data/tp_04/fusion.txt", "r", encoding="utf-8") as f:
    print(f.read())
    
# Afficher la liste des fichiers du dossier
print(os.listdir("data/tp_04"))
````
