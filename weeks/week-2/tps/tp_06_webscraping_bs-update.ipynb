## 1. Balises HTML


En HTML, on utilise des balises pour séparer les différents morceaux du code qui correspondent à des éléments sur la page. Par exemple ``<head>`` est la balise pour le header du fichier html, ``<title>`` pour son titre, ``<p>`` pour un pragraphe, ``<h1>``, ``<h2>``, ``<h3>``,... pour des headlines ou ``<br>``pour un retour à la ligne.


Une balise est toujours fermée par la même balise avec un ``/``en plus. Par exemple : ``<title> Titre </title>``

#### Exemple : les balise des tableaux

$$
\begin{array}{rr} \hline
Balise  & \text{Description} \\ \hline
< table> & \text{Tableau} \\
< caption>& \text{Titre du tableau} \\
< tr> & \text{Ligne de tableau} \\
< th> & \text{Cellule d'en-tête}\\
< td> & \text{Cellule} \\
< thead> & \text{Section de l'en-tête du tableau} \\
< tbody> & \text{Section du corps du tableau} \\
< tfoot> & \text{Section du pied du tableau} \\
\end{array}
$$

## 2. Parent et enfant

Dans le cadre du langage HTML, les termes de parents (parent) et enfants (child) servent à désigner des élements emboîtés les uns dans les autres. Dans la construction suivante, par exemple :


Sur la page web, cela apparaitra de la manière suivante : 

<div> 
    <p>
       bla,bla
    </p>
</div>

On dira que l'élément ``<div>`` est le parent de l'élément ``<p>`` tandis que l'élément ``<p>`` est l'enfant de l'élément ``<div>``.

## Exercice 1 : obtenir la liste des équipes de Ligue 1

Sur la page https://fr.wikipedia.org/wiki/Championnat_de_France_de_football_2019-2020, on veut scraper le tableau de la section Participants (celui avec ces colonnes : (Club, Dernière montée, Budget en M€, ...)

#### 1) Trouver le tableau


```python
import urllib
import bs4
import requests

# A toi de jouer
url="https://fr.wikipedia.org/wiki/Championnat_de_France_de_football_2019-2020"
headers = {
    "User-Agent": "unnomlambda/1.0 (shikamaru--@hotmail.fr)"
}

response = requests.get(url, headers=headers)
print(f"Status code: {response.status_code}")

wikitables = page.find_all("table", class_='wikitable sortable')
print(f"il y a: {len(wikitables)} tables")

#vérifier le tableau qu'on a besoin :
for x, table in enumerate (wikitables):
   if "Dernière" "montée" in table.text:
       print("index:", x)
```

#### 2) Récupérer chaque ligne du table


```python
# A toi de jouer
#Extraire les en-têtes (th)
tete_html = wikitables[0].find_all("th")

tete = []
for th in tete_html:
     tete.append(th.text.rstrip("\n"))
tete

#Extraire les lines:
# On ignore la première ligne (les en-têtes déjà récupérés)
lines_html =wikitables[0].find_all("tr")
lines = []
for tr in lines_html[1:]:
    line = []
    tds = tr.find_all("td")
    for td in tds:
        line.append(td.text.rstrip("\n"))
    lines.append(line)

lines

import pandas as pd

df = pd.DataFrame(lines, columns= tete)
df

#for i , element in enumerate(lines):
#    print(f"lines{i}: ", element,"\n")

```

## Exercice 2
- A partir de la suivante : [pokedex](https://pokemondb.net/pokedex/national), l'objectif est de constituer la liste de tous les pokemons
 - 1. Dans un premier temps, on voudrait la liste avec nom, numéro et types pour chaque pokémon
 - 2. Ensuite, on souhaiterait aussi garder l'url pour chaque pokemon (par exemple en utilisant l'attribut ``href``) pour pouvoir ensuite faire une requête ``GET``sur cette page (par exemple : https://pokemondb.net/pokedex/venusaur) et récupérer les données de taille et de poids

```python
# A toi de jouer
import urllib
import bs4
import requests

url_pokedex = "https://pokemondb.net/pokedex/national"
headers = {
    "User-Agent": "unnomlambda/1.0 (shikamaru--@hotmail.fr)"
}

response = requests.get(url_pokedex, headers=headers)
print(f"Status code: {response.status_code}")
```

```python
HOST_URL = "https://pokemondb.net"


text = response.text
# Etape 2 : utiliser le package BeautifulSoup
page = bs4.BeautifulSoup(text, "html")

pokemons= []
cards = page.find_all("div", class_ = "infocard")
for card in cards:
      # Numéro
    numero = card.find("small").text.lstrip("#")
    # Nom et URL
    link = card.find("a", class_ = "ent-name")
    nom = link.text
    url_poke = link.get("href")
   
    #type:
    types = [t.text for t in card.find_all("a", class_ = "itype")]
   
    pokemon = {
        "Numéro ": numero,
        "Name": nom,
        "Type":" ".join(types),    
        "url":HOST_URL + url_poke
    }
    pokemons.append(pokemon)
```

```python
df = pd.DataFrame(pokemons)
print(f"Liste initiale de {len(df)} Pokémon extraite.")
df
```

```python
#j'ai récupérer tous les liens de ppokemons
liens = []
tailles = []
poids = []
for pokemon in pokemons:
    lien = pokemon["url"]
    headers = {
    "User-Agent": "unnomlambda/1.0 (shikamaru--@hotmail.fr)"
    }
    rep = requests.get(lien, headers=headers)
    liens.append(lien)
#for l in liens: print(l) /afficher tous les liens 
    soup_poke = bs4.BeautifulSoup(rep.text, "html")
    vitals = soup_poke.find("table", {"class": "vitals-table"})

# Extraction du texte brut
    h_text = vitals.find("th", string="Height").find_next_sibling("td").text
    w_text = vitals.find("th", string="Weight").find_next_sibling("td").text
        
# Nettoyage pour ne garder que la valeur (ex: 0.7)
    h_clean = h_text.split("m")[0].strip()
    w_clean = w_text.split("kg")[0].strip()
        
    tailles.append(h_clean)
    poids.append(w_clean) 
```

```python
# 3. Ajout au DataFrame final
df["Taille (m)"] = tailles
df["Poids (kg)"] = poids


# Affiche les 10 premières lignes pour vérifier que Taille et Poids sont bien là
print(df.head(10))
```
