## 1. Balises HTML


En HTML, on utilise des balises pour séparer les différents morceaux du code qui correspondent à des éléments sur la page. Par exemple ``<head>`` est la balise pour le header du fichier html, ``<title>`` pour son titre, ``<p>`` pour un pragraphe, ``<h1>``, ``<h2>``, ``<h3>``,... pour des headlines ou ``<br>``pour un retour à la ligne.


Une balise est toujours fermée par la même balise avec un ``/``en plus. Par exemple : ``<title> Titre </title>``

#### Exemple : les balise des tableaux

$$
\begin{array}{rr} \hline
Balise  & \text{Description} \\ \hline
< table> & \text{Tableau} \\
< caption>& \text{Titre du tableau} \\
< tr> & \text{Ligne de tableau} \\
< th> & \text{Cellule d'en-tête}\\
< td> & \text{Cellule} \\
< thead> & \text{Section de l'en-tête du tableau} \\
< tbody> & \text{Section du corps du tableau} \\
< tfoot> & \text{Section du pied du tableau} \\
\end{array}
$$

##### Application : un tableau en HTML

## 2. Parent et enfant

Dans le cadre du langage HTML, les termes de parents (parent) et enfants (child) servent à désigner des élements emboîtés les uns dans les autres. Dans la construction suivante, par exemple :


<div> 
    <p>
       bla,bla
    </p>
</div>



<div> 

</div>

<p>
   bla,bla
</p>

Sur la page web, cela apparaitra de la manière suivante : 

<div> 
    <p>
       bla,bla
    </p>
</div>

On dira que l'élément ``<div>`` est le parent de l'élément ``<p>`` tandis que l'élément ``<p>`` est l'enfant de l'élément ``<div>``.

### Méthode ``.find_all()``
Elle renvoie toutes les occurences trouvées avec cette balise 

```python
all_tables = page.find_all("table")
print("Il y a", len(all_tables), "éléments dans la page qui sont des <table>")
print(type(all_tables[0]))
```

### Attribut ``.text``
Il contient le texte entre des bablises

```python
paragraph = page.find_all("p")[10]
print(paragraph)
print(paragraph.text)
```

### Filtrer sur une classe HTML avec ``find``ou ``find_all``

```python
wikitables = page.find_all("table", class_='wikitable sortable')
print(len(wikitables), "objects with class wikitable sortable")
```

### ``.get``pour récupérer un attribut
Ici, on récupère l'attribut ``href``d'un tag avec ``.get("href")``

```python
wikitables = page.find_all("table", class_='wikitable sortable')
print(len(wikitables), "objects with class wikitable sortable")

# th indique la ligne d'entête du tableau, on prend la troisième cellule 
# de la liste des entêtes
tag_header = wikitables[0].find_all("th")[3]
print(tag_header)
print('\n')

# on cherche maintenant la balise a dedans
tag_a = tag_header.find("a")
print(tag_a)
print(tag_a.get("href"))
```

## Exercice 1 : obtenir la liste des équipes de Ligue 1

Sur la page https://fr.wikipedia.org/wiki/Championnat_de_France_de_football_2019-2020, on veut scraper le tableau de la section Participants (celui avec ces colonnes : (Club, Dernière montée, Budget en M€, ...)

#### 1) Trouver le tableau


```python
import requests
import bs4
import pandas as pd


url = "https://fr.wikipedia.org/wiki/Championnat_de_France_de_football_2019-2020"
headers = {
    "User-Agent": "unnomlambda/1.0 (votreadressemail)"
}

response = requests.get(url, headers=headers)
page = bs4.BeautifulSoup(response.text, "html.parser")

# 1. Identifier le tableau
# Sur cette page, le tableau des participants est le premier tableau de classe "wikitable"
tableau = page.find("table", {"class": "wikitable"})


#2) Récupérer chaque ligne du table

# 2. Extraire les en-têtes (th)
headers_table = [th.text.strip() for th in tableau.find_all("th")]

# 3. Extraire les lignes de données (tr)
data = []
rows = tableau.find_all("tr")

# On ignore la première ligne (les en-têtes déjà récupérés)
for row in rows[1:]:
    cells = row.find_all("td")
    # On vérifie que la ligne contient bien des données
    if cells:
        row_data = [cell.text.strip() for cell in cells]
        data.append(row_data)

# 4. Création du DataFrame
df = pd.DataFrame(data, columns=headers_table[:len(data[0])])

# Affichage du résultat
print(df.head())
```

```python
import requests
import bs4
import pandas as pd

url = "https://fr.wikipedia.org/wiki/Championnat_de_France_de_football_2019-2020"
headers = {
    "User-Agent": "unnomlambda/1.0 (votreadressemail)"

}

response = requests.get(url, headers=headers)
page = bs4.BeautifulSoup(response.text, "html")

# Trouver tous les tableaux
tables = page.find_all("table")

table_participants = None

for table in tables:
    # récupérer la première ligne
    first_row = table.find("tr")
    if not first_row:
        continue

    headers = [c.get_text(strip=True) for c in first_row.find_all(["th", "td"])]

    # Vérifier si c'est le bon tableau
    if len(headers) > 0 and headers[0] == "Club":
        table_participants = table
        break

# Vérification
print("Tableau trouvé :", table_participants is not None)


#2) Récupérer chaque ligne du table

# Extraire les lignes
rows = table_participants.find_all("tr")

# Colonnes
colonnes = [c.get_text(strip=True) for c in rows[0].find_all(["th", "td"])]

# Données
donnees = []
for row in rows[1:]:
    cells = row.find_all("td")
    if len(cells) == len(colonnes):
        donnees.append([c.get_text(strip=True) for c in cells])

# DataFrame
df = pd.DataFrame(donnees, columns=colonnes)[:5]
df
```

## Exercice 2
- A partir de la suivante : [pokedex](https://pokemondb.net/pokedex/national), l'objectif est de constituer la liste de tous les pokemons
 - 1. Dans un premier temps, on voudrait la liste avec nom, numéro et types pour chaque pokémon
 - 2. Ensuite, on souhaiterait aussi garder l'url pour chaque pokemon (par exemple en utilisant l'attribut ``href``) pour pouvoir ensuite faire une requête ``GET``sur cette page (par exemple : https://pokemondb.net/pokedex/venusaur) et récupérer les données de taille et de poids

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time


# URL spécifique demandée
url_base = "https://pokemondb.net"
url_pokedex = f"{url_base}/pokedex/national"

headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url_pokedex, headers=headers)
soup = bs4.BeautifulSoup(response.text, "html.parser")

# 1. Extraction de la liste complète (Nom, Numéro, Types, URL)
pokemons = []
# Sur /national, les cartes sont dans des éléments avec la classe "infocard"
cards = soup.find_all("div", {"class": "infocard"})

for card in cards:
    # Numéro
    numero = card.find("small").text.strip("#")
    # Nom et URL
    link = card.find("a", {"class": "ent-name"})
    nom = link.text
    url_poke = url_base + link["href"]
    # Types
    # Les types sont dans les petits liens après le nom
    types = [t.text for t in card.find_all("a", {"class": "feat-icon"})]
    
    pokemons.append({
        "Numéro": numero,
        "Nom": nom,
        "Types": ", ".join(types),
        "URL": url_poke
    })

df = pd.DataFrame(pokemons)
print(f"Liste initiale de {len(df)} Pokémon extraite.")

# 2. Récupération de la taille et du poids pour chaque page
tailles, poids = [], []

print("Récupération des détails (taille/poids)...")

for url in df["URL"]:
    try:
        rep = requests.get(url, headers=headers)
        soup_poke = bs4.BeautifulSoup(rep.text, "html.parser")
        vitals = soup_poke.find("table", {"class": "vitals-table"})
        
        # Extraction du texte brut
        h_text = vitals.find("th", string="Height").find_next_sibling("td").text
        w_text = vitals.find("th", string="Weight").find_next_sibling("td").text
        
        # Nettoyage pour ne garder que la valeur (ex: 0.7)
        h_clean = h_text.split("m")[0].strip()
        w_clean = w_text.split("kg")[0].strip()
        
        tailles.append(h_clean)
        poids.append(w_clean)
        
    except:
        tailles.append(None)
        poids.append(None)
        
    time.sleep(0.1) # Petite pause pour le serveur

# 3. Ajout au DataFrame final
df["Taille (m)"] = tailles
df["Poids (kg)"] = poids


# Affiche les 10 premières lignes pour vérifier que Taille et Poids sont bien là
print(df.head(10))

# Sauvegardez le résultat pour ne plus avoir à relancer le scraping
df.to_csv("pokedex_final.csv", index=False)
print("Données sauvegardées dans 'pokedex_final.csv'")
```
